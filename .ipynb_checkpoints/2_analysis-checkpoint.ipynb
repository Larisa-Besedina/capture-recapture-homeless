6. правило триангуляции и итоговая таблица;
7. графики.

Все параметры анализа (`SEED`, `N_BOOT`, пороги отбора моделей, пути)
задаются в `config.py` — не меняйте их в ячейках.

## 3. Описательная статистика

Ключевые диагностики перед оценкой численности:

* `f1_share` — доля людей, попавших ровно в один список. Чем она выше, тем
  сильнее оценка зависит от предположений модели.
* `f1_f2` — отношение f1/f2, на нём напрямую строятся оценки Чао и Зельтермана.
  Пустое значение означало бы f2 = 0; в исходных данных такого нет, минимум
  f2 = 4.
* `*_ovl` — доля меньшего источника, покрытая большим (мера зависимости списков).

```python
df_metrics = descriptives.all_strata_metrics(raw_tables)
data_io.save_table(df_metrics, config.STRATA_METRICS)
descriptives.format_metrics(df_metrics)
```

```python
descriptives.metrics_summary(df_metrics)
```

## 4. Лог-линейные модели

Схема оценки: пуассоновская лог-линейная модель подгоняется по наблюдаемым
ячейкам таблицы сопряжённости (все, кроме (0,0,0)), затем предсказанное
значение для (0,0,0) в каждой страте берётся как оценка ненаблюдённых.

Перебираются все иерархические модели с парными взаимодействиями между
источниками и между источниками и ковариатами. Из них:

* **насыщенная** — модель с полным набором парных взаимодействий между
  источниками (даёт верхнюю границу оценки);
* **ненасыщенная** — лучшая по BIC модель без полного набора (нижняя граница).

Модели отбрасываются, если IRLS не сошёлся или `deviance/df > 3`.
Доверительные интервалы — параметрический бутстреп: из подогнанных
ожидаемых значений генерируются пуассоновские counts, модель
переподгоняется, берутся квантили распределения N.

Важная оговорка для текста статьи: интервал условен на выбранной модели и
не учитывает неопределённость самого выбора среди ~1000 моделей, поэтому
реальная неопределённость больше номинальных 95%.

Ячейка ниже — самая долгая в ноутбуке (порядка 1000 моделей на сезон плюс
бутстреп).

```python
loglin = pipeline.run_loglinear(raw_tables, n_boot=config.N_BOOT, verbose=True)

data_io.save_table(loglin["models"], config.LOGLIN_MODEL_SUMMARY)
data_io.save_table(loglin["coefficients"], config.LOGLIN_COEFFICIENTS)
```

```python
# Сводка по выбранным моделям
cols = ["season", "method", "model", "bic", "deviance", "df_resid",
        "obs", "unobs", "est", "ci_lower", "ci_upper", "coverage_%",
        "boot_success_share", "n_models_considered"]
loglin["models"][cols].round(1)
```

## 5. Оценки Чао и Зельтермана

Обе оценки используют только частоты захвата f1 и f2 внутри страты, то есть не
требуют предположения о независимости списков, но чувствительны к малым f2.

Основной расчёт — классическая формула Чао `n + f1²/(2·f2)`
(`config.CHAO_BIAS_CORRECTED = False`). Она применима, потому что f2 ≥ 4 во
всех стратах, и воспроизводима читателем по описательной таблице выше.
Если в новых данных появится страта с f2 < 3, расчёт выдаст предупреждение.

Что важно в реализации:

* при f2\* = 0 в ресемпле оценка **не подменяется на n**. Такая подмена
  переворачивала смысл: f2 → 0 означает «перекрытия почти нет, скрытых очень
  много», а записывалось как «скрытых нет». Теперь такие итерации
  исключаются из перцентилей, а их доля выводится в колонке
  `degenerate_share`;
* агрегаты (пол, total) считаются как сумма страт внутри итерации; итерация
  исключается, если вырождена хотя бы одна входящая страта — поэтому у
  агрегатов `degenerate_share` выше, чем у отдельных страт;
* для страт с малым f2 верхняя граница интервала определяется ресемплами с
  f2\* = 1 и содержательно не информативна. Это свойство самого оценщика, а
  не ошибка: имейте в виду при интерпретации.

### Проверка устойчивости: две формулы Чао рядом

Таблица показывает, где выбор формулы имеет значение. На итогах по сезонам
расхождение около 2–3%, на маленьких стратах доходит до 20%. Колонка
`f1_f2_outlier` помечает страты, где отношение f1/f2 выбивается из общего
ряда: там оценка ненаблюдённых опирается на единицы человек в пересечении
списков и заслуживает отдельного разбора.

```python
chao_variants = chao_zelterman.chao_variant_comparison(raw_tables)
data_io.save_table(chao_variants, config.CHAO_VARIANTS)
chao_variants.round(2)
```

```python
cz = pipeline.run_chao_zelterman(raw_tables, n_boot=config.N_BOOT)
cz.query("strata == 'total'").round(1)
```

## 7. Правило триангуляции

Три метода дают три разные оценки, и правило сводит их к одной:

* оценка Чао ниже ненасыщенной лог-линейной → берём ненасыщенную;
* оценка Чао выше насыщенной → берём насыщенную;
* иначе оставляем Чао.

Это эвристика, а не статистическая процедура: итоговая оценка не является
оценкой какой-либо одной модели, и номинальное покрытие её интервала не
гарантировано. Правило предполагает, что ненасыщенная модель даёт нижнюю
границу, а насыщенная — верхнюю; строки, где это нарушено, помечаются
флагом `corridor_invalid`, и их нужно разобрать вручную.

```python
df_final = triangulation.apply_triangulation(all_estimates)
data_io.save_table(df_final, config.FINAL_ESTIMATES)

df_final[["year_season", "strata", "obs", "unobs", "est",
          "ci_lower", "ci_upper", "coverage_%", "method_ru"]].round(1)
```

```python
# Проверка нарушений предпосылки правила
df_final[df_final["corridor_invalid"]][
    ["year_season", "strata", "chao_est", "nonsat_est", "sat_est"]
]
```
